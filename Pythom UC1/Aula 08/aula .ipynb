```python
def fatorial(n):
    if n <0:
        return "Numero inválido para fatorial."
    elif n == 0:
        return 1
    else:
        resultado =1 
        for i in range(1, n+1):
            resultado *=i
        return resultado
#Exemplo de chamada
print("Fatorial de 5:", fatorial(5))
```

```python
def eh_palindromo(palavra):
    return palavra == palavra[::-1]
# Exemplos de chamada
print("ana é palíndromo?", eh_palindromo("ana"))
print("python é palíndromo?", eh_palindromo("python"))
```

exerc 08

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1
    return contador
# Exemplo de chamada
print("Número de vogais em 'Python':", contar_vogais("Python"))
```

exerc 09

```python
def tabuada(n):
    resultado = []
    for i in range(1, 11):
        resultado.append(n * i)
    return resultado
# Exemplo de chamada
print("Tabuada do 5:", tabuada(5))
```

```python
def ordenar_por_segundo(lista_tuplas):
    return sorted(lista_tuplas, key=lambda x: x[1])
# Exemplo de chamada
lista_exemplo = [("a", 3), ("b", 1), ("c", 2)]
print("Lista ordenada:", ordenar_por_segundo(lista_exemplo))
```
