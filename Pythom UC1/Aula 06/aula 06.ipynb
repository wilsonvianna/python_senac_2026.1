WHILE

```python
contador=0
while(contador<=5):
    print(f"O valor do contador é.:{contador}:.")
    contador+=1
print("fim!!!!")
```

Matriz

```python
matriz=[
    [1,2,3],
    ['A','B','C'],
    [4,"D',5}"]
]
print(matriz)
```

```python
for linha in matriz:
    print("[", end="")
    for elemento in linha:
       
     print(f"{elemento}, ",end="")
     print("]", end="")
print("Fim!!")
```

```python
for i in range(10):
    if i == 5:
        break #para o loop quando i for 5
    if i %2 == 0:
        continue #Pula numeros pares
    print(i)
```
